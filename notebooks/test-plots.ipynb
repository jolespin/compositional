```python
import numpy as np
import compositional as coda
import pandas as pd
from scipy.stats import nbinom
```

```python
# Generate synthetic data with a high zero fraction using the Negative Binomial distribution
mu, alpha = 13, 21  # Higher alpha for more zeros
var = mu + alpha * mu**2
p = mu / var
n = mu * p

X_values = nbinom.rvs(n=n, p=p, size=(100,1000), random_state=42)
X = pd.DataFrame(X_values)
X.index = [f"Sample_{i}" for i in range(X.shape[0])]
X.columns = [f"Feature_{j}" for j in range(X.shape[1])]
print(f"Zero fraction ({n}, {p}): {np.mean(X == 0):.4f}")
X.head()
```

```python
fig, ax, im = coda.plot_compositions(X)
fig
```

```python
fig, ax, im = coda.plot_prevalence(X)
fig
```
